```python
class Solution:
    def containsDuplicate(self, nums: list[int]) -> bool:
        freq = {}
        for i in nums:
            if i in freq:
                return True
            else:
                freq[i] = 1

        return False
```

```python
class Solution:
    def containsNearbyDuplicate(self, nums: list[int], k: int) -> bool:
        freq = {}
        for i in range(len(nums)):
            if nums[i] in freq:
                if i - freq[nums[i]] <= k:
                    return True

            freq[nums[i]] = i

        return False
```

```python
class Solution:
    def firstRepeating(self, arr: list[int]) -> int:
        freq = {}
        for i in arr:
            if i in freq:
                freq[i] += 1
            else:
                freq[i] = 1

        for i in arr:
            if freq[i] > 1:
                return i

        return -1
```

```python
class Solution:
    def firstNonRepeating(self, arr: list[int]) -> int:
        freq = {}
        for i in arr:
            if i in freq:
                freq[i] += 1
            else:
                freq[i] = 1

        for i in arr:
            if freq[i] == 1:
                return i

        return -1
```

```python
class Solution:
    def majorityElement(self, nums: list[int]) -> int:
        freq = {}
        for i in nums:
            if i in freq:
                freq[i] += 1
            else:
                freq[i] = 1

        for i in freq:
            if freq[i] > len(nums) // 2:
                return i

        return -1
```

```python
class Solution:
    def findDuplicates(self, nums: list[int]) -> list[int]:
        freq = {}
        for i in nums:
            if i in freq:
                freq[i] += 1
            else:
                freq[i] = 1

        res = []
        for i in freq:
            if freq[i] == 2:
                res.append(i)

        return res
```
